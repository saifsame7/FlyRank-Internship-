## 1. My lane as an ML task (type)

**Ranking** — this is a content-refresh ranking problem. Given 30,000 pseudonymized content items across 32 clients, the goal is to surface the items most likely needing editorial attention (declining traffic/engagement). The output is a ranked queue telling reviewers *what to look at first*.

```python
print('Task type: Ranking — content refresh ranking problem')
```

## 2. Target or proxy

**Target:** `is_declining_label` = (`trend_direction` == "down"). This binary label indicates whether a content item's traffic is trending downward.

**Proxy source:** Derived from `trend_direction` column, which itself is computed from `trend_pct`. **Critical: `trend_direction` and `trend_pct` are NEVER valid features** — they would be label leakage since the label is derived from them.

**Alternative proxies that can be used as features:** `engagement_rate`, `scroll_rate`, `ai_traffic_pct`, `ctr`, client-specific baselines, `content_type` encoded flags.

```python
print('Target is is_declining_label = (trend_direction == "down") — never use trend_direction or trend_pct as features')
```

## 3. Success metric

**Precision@50** — what percentage of the top-50 items reviewed are actually declining. The baseline (hand-rule) achieves ~0.240; a well-trained model can reach 0.68–0.74, representing ~3x lift.

**Why this metric:** Directly maps to editorial efficiency — reviewing fewer pages to catch the same number of declining items. A claim of "0.70 Precision@50" means roughly 35 of the top-50 reviewed items will need attention.

**Directional wording for report:** "observed Precision@50 of 0.70 on the test split, compared to 0.240 for the hand-rule baseline."

```python
print('Success metric: Precision@50 - baseline 0.240, model target 0.68-0.74 (~3x lift)')
```

## 4. The unit of analysis, as a real dataframe

**One row = one pseudonymized content item.**

```python
import pandas as pd
df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
print('Shape:', df.shape)
print('One row = one pseudonymized content item')
print('trend_direction distribution:')
print(df['trend_direction'].value_counts())
print()
print('Derived is_declining_label:')
print((df['trend_direction'] == 'down').value_counts())
print('Note: is_declining_label is NOT a stored column; it is derived from trend_direction')
```

## 5. Why ML beats a fixed rule here

**Patterns too messy for an if-statement:**
- Missingness follows content_type (blind fillna injects category signal)
- Rate columns are ×100 percentages (easy to misinterpret: 0.76 = 0.76%, not 76%)
- 5 trend_direction values vs. binary decision
- Client-specific history depth varies wildly (some clients have little usable history)
- `avg_position = 0` means "no data" in 1,205 rows
- ML model learns feature interactions automatically; rules require manual case analysis for every edge case

```python
print('Why ML: automates pattern detection across messy real-world data that fixed rules cannot handle')
```
